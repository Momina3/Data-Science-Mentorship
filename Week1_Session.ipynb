```python
import seaborn as sns
import matplotlib.pyplot as plt
#barplot
sns.barplot(
    data=df,
    x="sex",
    y="survived"
)

plt.title("survival rate by gender")
plt.xlabel("gender")
plt.ylabel("survival rate")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
#barplot
sns.barplot(
    data=df,
    x="age",
    y="survived"
)

plt.title("survival rate by age")
plt.xlabel("age")
plt.ylabel("survival rate")
plt.show()
```

```python
#histogram plot
sns.histplot(
    data=df,
    x="age",
    bins=20
)
plt.title("age distribution")
plt.xlabel("age")
plt.ylabel("count of passengers")
plt.show()
```
